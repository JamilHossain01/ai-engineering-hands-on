```python
from openai import OpenAI
# from IPython.display import Markdown, display
```

```python
client = OpenAI(
    base_url = "http://localhost:11434/v1",
    api_key ="ollama"
)
```

```python
MODEL_A = "qwen3:8b"
MODEL_B = "deepseek-coder:6.7b"
```

```python
initial_prompt = "Write a Python function to check if a string is a palindrome and explain its time complexity."
print(f"👤 [User]: {initial_prompt}\n" + "="*60)
```

```python
response = client.chat.completions.create(
    model = MODEL_A,
    messages = [{"role":"user","content":initial_prompt}]
)
```

```python
answer_a = response.choices[0].message.content
print (f"[Model A -{MODEL_A}]:\n{answer_a}\n" + "="*60)
```

```python
propmt_for_b = f"Review the following code and provide feedback on its correctness,efficienty, and any potential improvement suggestions:\n{answer_a} "
```

```python
response_b = client.chat.completions.create(
    model = MODEL_B,
    messages = [{"role":"user","content": propmt_for_b}]
)
answer_b = response_b.choices[0].message.content
print (f"Model B - {MODEL_B}:\n{answer_b}\n" + "="*60)
```
