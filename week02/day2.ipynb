```python
def ask_ai(propmt,model="qwen3:8b"):
    response = ollama.chat(
        model = model,
        messages = [
            {
                "role": "user",
                "content": propmt
            }
        ]
    )
    return response ["message"] ["content"]
```

```python
ask_ai("Explain Flutter Clean Architecture in simple language.")
```

```python
ai_demo = gr.Interface(
    fn=ask_ai,
    inputs="textbox",
    outputs="textbox",
    title="my local ai assistant",
)
```

```python
stream = ollama.chat(
    model="qwen3:8b",
    messages=[
        {
            "role": "user",
            "content": "Explain RAG in simple language.",
        }
    ],
    stream=True,
)

result = ""

for chunk in stream:
    content = chunk["message"]["content"]
    result += content

print(result)
```

```python
def stream_ai(propmt,model="qwen3:8b"):
    stream = ollama.chat(
        model=model,
        messages=[
            {
                "role":"user",
                "content":propmt,
            }
        ],
        stream=True,
    )
    result = ""
    for chunk in stream:
        content = chunk["message"]["content"]
        result += content
        yield result
```

```python
result = stream_ai(
    "Explain AI Engineer Fundamnetal in simple language."
)
```

```python
for text in result:
    print(text)
```

```python
stream_demo = gr.Interface(
    fn=stream_ai,
    inputs="textbox",
    outputs="markdown",
    title="Streaming Local AI Assistant",
    description="Chat with Qwen3 using ollama"
)
```

```python
stream_demo.launch()
```

```python
def fetch_website_contents(url):
    response = requests.get(url)

    soup = BeautifulSoup(
        response.text,
        "html.parser"
    )

    text = soup.get_text(
        separator=" ",
        strip=True
    )

    return text
```

```python
website_text = fetch_website_contents(
   "https://8bit.com.bd/careers#apply"
)
```

```python
print(website_text[:1000])
```

```python
def create_brochure(url):
    website_text = fetch_website_contents(url)

    prompt = f"""
You are an expert marketing copywriter.

Create a professional company brochure based only on
the information provided in the website content.

Include:

1. Company name
2. About the company
3. Main services
4. Products or solutions
5. Key strengths
6. Contact information if available

Use Markdown headings and bullet points.

Do not invent information that is not present
in the website content.

Website content:
{website_text}
"""

    return ask_ai(prompt, "qwen3:8b")
```

```python
brochure = create_brochure("https://8bit.com.bd/careers#apply")
```

```python
print(brochure)
```

```python
def create_brochure_stream(url):
    website_text = fetch_website_contents(url)
    propmt = f""" 
You are an expert marketing copywriter.
Create a professional company brochure based only on
the information provided in the website content.

Include:

1. Company name
2. About the company
3. Main services
4. Products or solutions
5. Key strengths
6. Contact information if available

Use Markdown headings and bullet points.

Do not invent information that is not present
in the website content.

Website content:
{website_text}
"""
    yield from stream_ai(propmt,"qwen3:8b")
```

```python
for text in create_brochure_stream("https://8bit.com.bd/careers#apply"):
    print(text, end="", flush=True)
```

```python
brochure_demo = gr.Interface(
    fn=create_brochure_stream,
    inputs=gr.Textbox(
        label="Company Website Url",
        placeholder="https://8bit.com.bd/careers#apply"
    ),
    outputs=gr.Markdown(
        label="Company Brochure"
    ),
    title="Ai Company Brochure generator",
    description="Enter a company website and genteate a company brochure using LLM"
)
```

```python
brochure_demo.launch()
```
